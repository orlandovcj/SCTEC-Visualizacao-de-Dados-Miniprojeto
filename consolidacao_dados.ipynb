## Consolidação dos dados

Este Notebook realiza a consolidação dos arquivos CSV em um único arquivo, facilitando a análise preliminar dos dados.

Nota: Inicialmente foram examinados os cabeçalhos dos arquivos CSV para verificar se todos possuem a mesma estrutura e tipos de dados. Em seguida, os arquivos foram concatenados em um único arquivo CSV, que será utilizado na etapa de análise preliminar.

```python
# Consolidar arquivos CSV em um único arquivo CSV
#  
import pandas as pd
import glob
import os

# Configurações
pasta_dados = 'Dados'
arquivo_saida = os.path.join(pasta_dados, 'BPS_20_26_OrlandoCastro.csv')
padrao_arquivos = os.path.join(pasta_dados, '*.csv')

def concatenar_csvs():
    # Lista todos os arquivos .csv na pasta
    arquivos = glob.glob(padrao_arquivos)
    
    if not arquivos:
        print("Nenhum arquivo .csv encontrado na pasta 'Dados'.")
        return

    # Lista para armazenar os dataframes
    lista_df = []

    for arquivo in arquivos:
        print(f"Lendo {arquivo}...")
        # Lendo o CSV com separador ';' conforme analisado nos cabeçalhos
        df = pd.read_csv(arquivo, sep=';', encoding='utf-8', dtype=str)
        lista_df.append(df)

    # Concatena todos os dataframes da lista
    df_final = pd.concat(lista_df, ignore_index=True)

    # Salva o resultado em um novo arquivo CSV
    # index=False evita que o pandas crie uma coluna de índice numérica no início
    df_final.to_csv(arquivo_saida, index=False, sep=';', encoding='utf-8')
    print(f"\nSucesso! Arquivo concatenado salvo como: {arquivo_saida}")

if __name__ == "__main__":
    concatenar_csvs()
```
